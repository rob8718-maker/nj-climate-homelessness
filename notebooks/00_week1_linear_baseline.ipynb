```python
import numpy as np
import matplotlib.pyplot as plt

def linear_model(x, m, b):
    return m * x + b

x = np.linspace(0, 10, 100)
y = linear_model(x, m=2, b=1)

plt.plot(x, y)
plt.title("f(x) = 2x + 1 — your first model")
plt.xlabel("x — input / feature")
plt.ylabel("f(x) — output / prediction")
plt.show()


%matplotlib inline

import matplotlib
print(matplotlib.__version__)



%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

def linear_model(x, m, b):
    return m * x + b

x = np.linspace(0, 10, 100)
y = linear_model(x, m=2, b=1)

plt.plot(x, y)
plt.title("f(x) = 2x + 1 — your first model")
plt.xlabel("x — input / feature")
plt.ylabel("f(x) — output / prediction")
plt.show()
```
